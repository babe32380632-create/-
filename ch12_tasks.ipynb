## 1. 載入 `inventory_turnovers.json` 為 `list` 並命名為 `inventory_turnovers_json`

````python
import json
with open("inventory_turnovers.json", "r") as f:
    inventory_turnovers_json = json.load(f)
````

````python
print(type(inventory_turnovers_json))
print(len(inventory_turnovers_json))
print(type(inventory_turnovers_json[0]))
print(len(inventory_turnovers_json[0]))
````

## 2. 將 `inventory_turnovers_json` 中所有資料期間 `inventory_turnover` 存貨周轉率任一期大於等於零的公司給予 `is_inventory_turnover_not_negative: True`，資料期間 `inventory_turnover`  存貨周轉率全部皆小於零的公司給予 `is_inventory_turnover_not_negative: False`，將判斷結果以 `dict` 紀錄後新增到 `list` 並命名為 `inventory_turnover_is_not_negative_list`


```
[
    {"ticker": "5566.TW",
     "is_inventory_turnover_not_negative": True},
    ...
    {"ticker": "5566.TWO",
     "is_inventory_turnover_not_negative": False}
]
```

````python
import pandas as pd
import numpy as np
inventory_turnover_is_not_negative_list = list()
for data in inventory_turnovers_json :
    ticker = data["ticker"]
    inventory_turnover = data["inventory_turnover"]
    if inventory_turnover is None : #整個存貨周轉沒有資料
        dict_to_append ={"ticker":ticker,
                         "is_inventory_turnover_not_negative":False}
        inventory_turnover_is_not_negative_list.append(dict_to_append)
    elif None in inventory_turnover.values(): #部分存貨沒有資料
        dict_to_append ={"ticker":ticker,
                         "is_inventory_turnover_not_negative":False}
        inventory_turnover_is_not_negative_list.append(dict_to_append)
    else :
        list_turnover = list(inventory_turnover.values())
        arr = np.array(list_turnover)
        delta = np.diff(arr)
        is_not_negative = delta >= 0
        is_not_negative_sum = is_not_negative.sum()
        dict_to_append ={"ticker":ticker,
                         "is_inventory_turnover_not_negative":bool(is_not_negative_sum)}
        inventory_turnover_is_not_negative_list.append(dict_to_append)
print(inventory_turnover_is_not_negative_list)
````

## 3. 將 `inventory_turnover_is_not_negative_list` 轉換為外型 `(200, 2)` 的資料框命名為 `inventory_turnover_is_not_negative_df`

````python
import pandas as pd
inventory_turnover_is_not_negative_df = pd.DataFrame(inventory_turnover_is_not_negative_list)
recreaseing_df = inventory_turnover_is_not_negative_df[
    inventory_turnover_is_not_negative_df["is_inventory_turnover_not_negative"] == True
]
print(recreaseing_df)
print("總家數：", len(recreaseing_df))

declining_df = inventory_turnover_is_not_negative_df[
    inventory_turnover_is_not_negative_df["is_inventory_turnover_not_negative"] == False
]
print(declining_df)
print("總家數：", len(declining_df))
print(inventory_turnover_is_not_negative_df)
````
