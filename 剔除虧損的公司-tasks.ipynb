## 1. 載入 `gross_profits_and_operating_margins.json` 為 `list` 並命名為 `gross_profits_and_operating_margins_json`

````python
import json
file_path = "gross_profits_and_operating_margins.json"
with open(file_path , "r") as f:
    gross_profits_and_operating_margins_json = json.load(f)
````

## 2. 將 `gross_profits_and_operating_margins_json` 中資料期間內有任一期 `net_income` 小於零的公司給予 `is_profitable: False`，資料期間內全部 `net_income` 大於零的公司給予 `is_profitable: True`，將判斷結果以 `dict` 紀錄後新增到 `list` 並命名為 `is_profitable_list`


```
[
    {"ticker": "5566.TW",
     "is_profitable": True},
    ...
    {"ticker": "5566.TWO",
     "is_profitable": False}
]
```

````python
import numpy as np
is_profitable_list = list()
for data in gross_profits_and_operating_margins_json:
    ticker = data["ticker"]
    net_income = data["net_income"]
    if net_income is None:
        dict_to_append = {"ticker":ticker,
                          "is_profitable":False}
        is_profitable_list.append(dict_to_append)
    else:
        net_income_values = list(net_income.values())
        arr = np.array(net_income_values) #將list轉成array
        arr_is_not_negative = arr >= 0  #轉成bool
        sum_arr = arr_is_not_negative.sum() #加總之後為0 ,is_profitable=Ture
        dict_to_appned = {"ticker":ticker,
                          "is_profitable":bool(sum_arr)}
        is_profitable_list.append(dict_to_appned)
print(is_profitable_list)
````

## 3. 將 `is_profitable_list` 轉換為外型 `(200, 2)` 的資料框命名為 `is_profitable_df`

````python
import pandas as pd
is_profitable_df = pd.DataFrame(is_profitable_list)
is_profitable_df
````
